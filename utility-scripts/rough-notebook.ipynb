# note: slug format observed:
1. dd-month_name-yyyy : Hindupur,Kanakapura,Vijayapura,Kyalanuru,Kollegal,Mysuru,Makavalli,Channapatna
2. dd-mm-yyyy : Ramanagara, Kolar


```python
months = ["0"+str(number) if number <10 else str(number) for number in range(1, 13)]
month_name = ["January", "February", "March", "April", "May", "June", "July", "August", "September", "October", "November", "December"]
days = ["0"+str(number) if number <10 else str(number) for number in range(1, 32)]
years = ["20"+str(number) for number in range(20, 27)]
valid_dates = []
for year in years:
    for month in months:
        for day in days:
            if month in ("01","03","05","07","08","10","12"):
                valid_dates.append(f"{day}-{month}-{year}")
            else:
                if month == "02":
                    if int(year) % 4 == 0:
                        if day in ["29"]:
                            valid_dates.append(f"{day}-{month}-{year}")
                    else:
                        if day in ["28"]:
                            valid_dates.append(f"{day}-{month}-{year}")

print(valid_dates)
```
